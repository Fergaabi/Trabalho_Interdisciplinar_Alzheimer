```python
# Ler o dataset

dados = pd.read_csv("alzheimer_dataset.csv")

dados.head()
```

```python
# Verificar dimensões

print(f"Linhas: {dados.shape[0]}")
print(f"Colunas: {dados.shape[1]}")
```

```python
# Mostrar nomes das colunas

print(dados.columns.tolist())
```

```python
# Informações gerais

dados.info()
```

```python
#Verificar valores nulos

dados.isnull().sum()
```

```python
# Estatísticas descritivas

dados.describe()
```

```python
# Distribuição da variável alvo

dados["Diagnosis"].value_counts(normalize=True) * 100
```

```python
# Verificar duplicados

dados.duplicated().sum()
```

```python
# Remover colunas inúteis

dados = dados.drop(
    columns=[
        "PatientID",
        "DoctorInCharge"
    ]
)

dados.head()
```

```python
# Conferir quantas colunas ficaram

print(dados.shape)
```

```python
# Separar atributos e alvo

X = dados.drop("Diagnosis", axis=1)

y = dados["Diagnosis"]

print(X.shape)
print(y.shape)
```

```python
# Verificar balanceamento

print(y.value_counts())
```

```python
# Salvar dataset preparado

dados.to_csv(
    "alzheimer_dataset_preparado.csv",
    index=False
)
```
